```python
# PARKING ANALYSIS
```

```python
#1. What is the total number of parking transactions?
parking.count()
```

```python
#2. What is the total parking revenue generated?
total_revenue = parking["Parking_Fee"].sum()
print("Total Revenue:", total_revenue)
```

```python
#3. What is the average parking fee?
parking_fee = parking["Parking_Fee"].mean()
print("Average_parking_fee :",parking_fee)
```

```python
#4. What is the average parking duration?
parking_duration = parking["Parking_Duration_Minutes"].mean()
print("Average_parking_duration :",parking_duration)
```

```python
# PAYMENT ANALYSIS
```

```python
#5. Which payment method is used the most?
parking["Payment_Method"].value_counts
```

```python
#6. Which payment method generates the highest revenue?
parking .groupby("Payment_Method")["Parking_Fee"].sum()
```

```python
# TIME ANALYSIS
```

```python
#7. Which time period has the highest number of transactions?
parking["Time_Period"].value_counts()
```

```python
#8. Which time period generates the highest revenue?
parking.groupby("Time_Period")["Parking_Fee"].sum()
```

```python
#9. Are there more transactions on weekdays or weekends?
parking["Day_Type"].value_counts()
```

```python
#10. Which day type generates higher revenue?
parking.groupby("Day_Type")["Parking_Fee"].sum()
```

```python
# Vehicle Analysis
```

```python
data = pd.merge(parking, vehicle, on="Vehicle_ID", how="inner")
```

```python
#11. Which vehicle type has the highest number of transactions?
vehicle_transaction = data["Vehicle_Type"].value_counts()
print(vehicle_transaction)
```

```python
#12. Which vehicle type generates the highest revenue?
data.groupby("Vehicle_Type")["Parking_Fee"].sum().sort_values(ascending=False)
```

```python
#13. Which vehicle type has the highest average parking duration?
data.groupby("Vehicle_Type")["Parking_Duration_Minutes"].mean()
```

```python
# Parking Slot & Monthly Analysis
```

```python
#20. Which parking slot is used the most?
parking["Parking_Slot"].value_counts()
```

```python
#21. How much revenue is generated in each month?
parking["Entry_Date"] = pd.to_datetime(parking["Entry_Date"])

monthly_revenue = parking.groupby(
    parking["Entry_Date"].dt.month
)["Parking_Fee"].sum()

print(monthly_revenue)
```

```python
#22. Which month generated the highest revenue?


parking["Entry_Date"] = pd.to_datetime(parking["Entry_Date"])
parking["Month"] = parking["Entry_Date"].dt.month_name()

monthly_revenue = parking.groupby("Month")["Parking_Fee"].sum()

print(monthly_revenue)
```
